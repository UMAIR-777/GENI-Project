```python
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
import time

# Set up ChromeDriver
chrome_driver_path = r"C:\Users\PMLS\Downloads\chromedriver-win64\chromedriver-win64\chromedriver.exe"
service = Service(chrome_driver_path)

# Initialize the WebDriver
driver = webdriver.Chrome(service=service)

# Function to extract user details (Name, City) from a LinkedIn profile
def extract_user_details(linkedin_url):
    driver.get(linkedin_url)
    
    # Wait for the page to load
    time.sleep(3)  # Increased time to allow elements to load
    
    try:
        # Extract Name (h1 tag)
        name = driver.find_element(By.TAG_NAME, "h1").text
        
        # Debugging: Print page source to ensure city info exists in the HTML
        print(driver.page_source)  # This will print all HTML, useful for debugging
        
        # Try to find the city element (span with class 'text-body-small inline t-black--light break-words')
        try:
            # Explicitly wait for the city element to be visible
            city_element = WebDriverWait(driver, 10).until(
                EC.presence_of_element_located((By.CSS_SELECTOR, "span.text-body-small.inline.t-black--light.break-words"))
            )
            city = city_element.text
        except Exception as e:
            city = "City element not found"
        
        user_details = {
            'Name': name,
            'City': city
        }
        return user_details
    except Exception as e:
        print(f"Error extracting details for {linkedin_url}: {e}")
        return None

# Function to extract details for multiple LinkedIn URLs dynamically
def extract_details_from_multiple_profiles(linkedin_urls):
    user_data = []
    for i, url in enumerate(linkedin_urls):
        print(f"Extracting data for URL {i+1}: {url}")
        details = extract_user_details(url)
        if details:
            user_data.append(details)
        # Wait for a few seconds before moving to the next profile
        time.sleep(2)
    
    # Return all extracted details
    return user_data

# Example usage: You can pass a list of LinkedIn URLs
linkedin_urls = [
    "https://www.linkedin.com/in/daryus-medora"
]

# Extract details for multiple LinkedIn profiles
user_data = extract_details_from_multiple_profiles(linkedin_urls)

# Display the extracted details
for user in user_data:
    print(user)

# Close the WebDriver
driver.quit()
```

```python
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.common.by import By
from selenium.webdriver.common.keys import Keys
import time

# Set up ChromeDriver
chrome_driver_path = r"C:\Users\PMLS\Downloads\chromedriver-win64\chromedriver-win64\chromedriver.exe"
service = Service(chrome_driver_path)

# Initialize the WebDriver
driver = webdriver.Chrome(service=service)

# Queries for various job titles and locations
queries = [
    {
        "title": "Software Engineer Profiles",
        "query": 'site:linkedin.com/in/ ("Software Engineer" OR "Software Developer") ("San Francisco" OR "New York" OR "Los Angeles" OR "Boston" OR "Seattle" OR "Austin" OR "Chicago" OR "Washington DC" OR "Denver" OR "Silicon Valley") ("United States")'
    },
    {
        "title": "Tech Lead Profiles",
        "query": 'site:linkedin.com/in/ ("Tech Lead") ("San Francisco" OR "New York" OR "Los Angeles" OR "Boston" OR "Seattle" OR "Austin" OR "Chicago" OR "Washington DC" OR "Denver" OR "Silicon Valley") ("United States")'
    },
    {
        "title": "Director Profiles",
        "query": 'site:linkedin.com/in/ ("Director") ("San Francisco" OR "New York" OR "Los Angeles" OR "Boston" OR "Seattle" OR "Austin" OR "Chicago" OR "Washington DC" OR "Denver" OR "Silicon Valley") ("United States")'
    },
    {
        "title": "Seed Investor Profiles",
        "query": 'site:linkedin.com/in/ ("Seed Investor") ("San Francisco" OR "New York" OR "Los Angeles" OR "Boston" OR "Seattle" OR "Austin" OR "Chicago" OR "Washington DC" OR "Denver" OR "Silicon Valley") ("United States")'
    },
    {
        "title": "Angel Investor Profiles",
        "query": 'site:linkedin.com/in/ ("Angel Investor") ("San Francisco" OR "New York" OR "Los Angeles" OR "Boston" OR "Seattle" OR "Austin" OR "Chicago" OR "Washington DC" OR "Denver" OR "Silicon Valley") ("United States")'
    }
]

# Dictionary to store categorized LinkedIn profile URLs
categorized_urls = {}

try:
    # Loop through all queries and fetch results from multiple pages
    for item in queries:
        title = item["title"]
        query = item["query"]
        categorized_urls[title] = []  # Initialize list for this category

        for page in range(5):  # Fetch results from 5 pages
            google_url = f"https://www.google.com/search?q={query}&start={page * 10}"
            
            # Open the URL
            driver.get(google_url)
            time.sleep(183)  # Wait for the page to load

            # Find all the result links
            search_results = driver.find_elements(By.CSS_SELECTOR, "a")
            
            # Collect LinkedIn URLs
            for result in search_results:
                url = result.get_attribute("href")
                # Filter LinkedIn profile URLs (contains 'linkedin.com/in/')
                if url and "linkedin.com/in/" in url:
                    categorized_urls[title].append(url)

            # Optional: Press the "Next" button instead of constructing URLs
            # Uncomment the lines below if you want to navigate using the "Next" button
            # try:
            #     next_button = driver.find_element(By.LINK_TEXT, "Next")
            #     next_button.click()
            #     time.sleep(3)
            # except Exception as e:
            #     print(f"No more pages for {title}")
            #     break

    # Remove duplicates and format URLs for display
    for category, urls in categorized_urls.items():
        categorized_urls[category] = sorted(set(urls))

    # Display categorized results
    for category, urls in categorized_urls.items():
        print(f"{category} Links:")
        print("-" * 60)
        if urls:
            for idx, url in enumerate(urls, start=1):
                print(f"{idx}. {url}")
        else:
            print("No URLs found.")
        print("\n")

finally:
    # Close the browser
    driver.quit()
```

```python
import pandas as pd
import os

# Assuming 'categorized_urls' is the dictionary containing URLs for each category
# For example: {'Software Engineer Profiles': [url1, url2], 'Tech Lead Profiles': [url3, url4], ...}

all_data = []

# Loop through each category and prepare the data with titles
for title, urls in categorized_urls.items():
    # Filter URLs that start with "https://www.linkedin.com/in"
    filtered_urls = [url for url in urls if url.startswith("https://www.linkedin.com/in")]
    
    # Add the title to each URL
    category_data = [{'Category': title, 'URL': url} for url in filtered_urls]
    
    # Append the category data to the all_data list
    all_data.extend(category_data)

# Create a DataFrame for all URLs and categories
df = pd.DataFrame(all_data)
# Generate a safe filename for the final CSV file
filename = "linkedin_profiles_with_categories.csv"

# Save the DataFrame to CSV
df.to_csv(filename, index=False)

print(f"All data saved to {filename}")
```
