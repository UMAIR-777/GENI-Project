```python
import requests
from typing import List, Optional
from langchain import LLMChain, PromptTemplate  # Hypothetical LangChain imports

# Example LLM initialization (using ChatGroq as provided)
from langchain_groq import ChatGroq
llm = ChatGroq(temperature=0, model="llama-3.3-70b-versatile")

# Function to fetch data from Google SERP API
def get_serp_data(keyword: str, result_count: int) -> dict:
    # Replace with actual API endpoint and authentication details
    serp_api_url = f"https://api.example.com/serp?query={keyword}&count={result_count}"
    response = requests.get(serp_api_url)
    response.raise_for_status()
    return response.json()

# Function to fetch keyword metrics from SEMRush
def get_semrush_data(keyword: str) -> dict:
    # Replace with actual API endpoint and authentication details
    semrush_api_url = f"https://api.example.com/semrush/keyword_overview?keyword={keyword}"
    response = requests.get(semrush_api_url)
    response.raise_for_status()
    return response.json()

# Build combined prompt for the LLM
def build_keyword_report_prompt(keyword: str, semrush_data: dict, serp_data: dict) -> str:
    prompt = f"""
Keyword: {keyword}

SEMRUSH DATA:
- Search Volume: {semrush_data.get('search_volume')}
- CPC: {semrush_data.get('cpc')}
- Competition: {semrush_data.get('competition')}
- Number of Results: {semrush_data.get('num_results')}
- Keyword Difficulty: {semrush_data.get('keyword_difficulty')}

SERP API RESULTS:
- Organic Results: {serp_data.get('organic_results')}
- Related Questions: {serp_data.get('related_questions')}
- Related Searches: {serp_data.get('related_searches')}
- Search Information: {serp_data.get('search_information')}

The provided content contains information from SEMRush and the SERP API for the keyword.
Your task is to analyze these results and create a 'Keyword Research Report'. Make it detailed, informative, and factual. Reference and include the SEMRush data, and ensure the report gives the end user comprehensive insight into this keyword.

Output just the report in Markdown (do not include markdown backticks) and nothing else.
"""
    return prompt

# Function to generate the keyword research report using the LLM
def generate_keyword_report(keyword: str, result_count: int) -> str:
    # Step 1: Retrieve data from the APIs
    serp_data = get_serp_data(keyword, result_count)
    semrush_data = get_semrush_data(keyword)
    
    # Step 2: Build the LLM prompt
    prompt = build_keyword_report_prompt(keyword, semrush_data, serp_data)
    
    # Step 3: Use LangChain to generate the report
    prompt_template = PromptTemplate(template=prompt, input_variables=[])
    chain = LLMChain(llm=llm, prompt=prompt_template)
    
    # Run the chain (no additional variables required as the prompt is pre-built)
    report = chain.run({})
    
    return report

# ---------------------------------------
# Example usage
# ---------------------------------------
if __name__ == "__main__":
    keyword = "social media marketing"
    result_count = 5  # Number of search results to analyze
    
    try:
        report = generate_keyword_report(keyword, result_count)
        print("----- Keyword Research Report -----\n")
        print(report)
    except Exception as e:
        print(f"An error occurred: {e}")
```
